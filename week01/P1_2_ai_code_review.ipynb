## Task 1: Write a Factorial Function

Write a Python function that computes the factorial of a non-negative integer. Start with this basic version:

````python
def factorial(n):
    """Compute the factorial of a non-negative integer n."""
    if n <= 1:
        return 1
    return n * factorial(n - 1)

# Test it
print(f"factorial(0) = {factorial(0)}")
print(f"factorial(1) = {factorial(1)}")
print(f"factorial(5) = {factorial(5)}")
print(f"factorial(10) = {factorial(10)}")
````

## Task 4: Implement the Improved Version

Based on your evaluation, write an improved version of the function:

````python
def factorial_improved(n: int):
    """
    Compute the factorial of a non-negative integer n.

    Parameters
    ----------
    n : int
        A non-negative integer.

    Returns
    -------
    int
        The factorial of n (n!).

    Raises
    ------
    ValueError
        If n is negative.
    TypeError
        If n is not an integer.
    """
    # error handling for invalid inputs
    if n < 0:
        raise ValueError("Input must be non-negative")
    if not isinstance(n, int):
        raise TypeError("Input must be an integer")

    # iterative implementation
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

# Test the improved version
test_cases = [0, 1, 5, 10]
for tc in test_cases:
    print(f"factorial_improved({tc}) = {factorial_improved(tc)}")

# Test edge cases
try:
    factorial_improved(-1)
except (ValueError, TypeError) as e:
    print(f"factorial_improved(-1) raised: {e}")

try:
    factorial_improved(3.5)
except (ValueError, TypeError) as e:
    print(f"factorial_improved(3.5) raised: {e}")
````

## Task 5: Verify Against the Standard Library

Python has a built-in factorial function. Use it to verify your implementation:

````python
import math

# Compare your function against math.factorial
test_values = [0, 1, 5, 10, 15, 20]
all_match = True

for n in test_values:
    yours = factorial_improved(n)
    correct = math.factorial(n)
    match = yours == correct
    if not match:
        all_match = False
    print(f"n={n:3d}: yours={yours:>20d}  correct={correct:>20d}  match={match}")

print(f"\nAll results match: {all_match}")
````
