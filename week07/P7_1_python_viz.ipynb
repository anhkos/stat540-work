## Task 1: Reproduce Two ggplot2 Charts in seaborn

Take two of your ggplot2 charts from R7.1 and recreate them in seaborn:

### Chart 1: Histogram (diamond prices)

````python
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.ticker import StrMethodFormatter

diamonds = sns.load_dataset("diamonds")

sns.set_theme(style="whitegrid")

fig, ax = plt.subplots(figsize=(8, 5))
sns.histplot(data=diamonds, x="price", binwidth=500,
             color="steelblue", edgecolor="white", ax=ax)
ax.xaxis.set_major_formatter(StrMethodFormatter("${x:,.0f}"))
ax.set_title("Diamond Prices Are Strongly Right-Skewed")
ax.set_xlabel("Price (USD)")
ax.set_ylabel("Number of diamonds")
plt.tight_layout()
plt.show()
````

### Chart 2: Boxplot (price by clarity, log scale)

````python
fig, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(data=diamonds, x="clarity", y="price", hue="clarity",
            palette="viridis", legend=False,
            flierprops={"alpha": 0.2, "markersize": 3}, ax=ax)
ax.set_yscale("log")
ax.yaxis.set_major_formatter(StrMethodFormatter("${x:,.0f}"))
ax.set_title("Price by Clarity (Log Scale)")
ax.set_xlabel("Clarity (worst to best)")
ax.set_ylabel("Price (USD, log scale)")
plt.tight_layout()
plt.show()
````

## Task 2: Reproduce One Chart in plotnine

Take one of your ggplot2 charts and reproduce it in plotnine:

````python
from plotnine import *

(
    ggplot(diamonds, aes(x="carat", y="price"))
    + geom_point(alpha=0.1, color="darkgreen")
    + facet_wrap("~cut", nrow=2)
    + scale_y_continuous(labels=lambda l: [f"${v:,.0f}" for v in l])
    + labs(
        title="Carat vs. Price Relationship Across Cuts",
        x="Carat",
        y="Price (USD)",
    )
    + theme_minimal()
    + theme(figure_size=(9, 6))
)
````

**Your turn:** How similar was the plotnine code to the ggplot2 code? What differences did you notice?

> Very similar the main difference is that plotnine is wrapped in a function, and has slightly different values for the labels in the `scale_y_continuous` function, but overall quite similar.
